```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableParallel
from langchain_community.document_loaders import TextLoader, PyPDFLoader
import os
```

```python
parser= StrOutputParser()
```

```python
os.environ['gemini_api'] = os.getenv('GEMINI_KEY', '')  
```

```python
llm=ChatGoogleGenerativeAI(
    model='gemini-3.5-flash',
    api_key=os.environ['gemini_api']
)
```

# LCEL-1- extracting the skills from the resume 

```python
skill_prompt= ChatPromptTemplate.from_messages([
    ('system','extract the skills from the resume.'),
    ('human','ressume: {resume}')
]

)

skill_chain=skill_prompt | llm | parser
```

# LCEL-2- summarize the project/ experience from the resume

```python
summarize_prompt= ChatPromptTemplate.from_messages([
    
    ('system','summarize the experience from the resume'),
    ('human','resume: {resume}')
    ]
)

summarize_chain=summarize_prompt | llm | parser
```

# LCEL-3- find the missing skills from the resume for data science roles

```python
missing_skill_prompt= ChatPromptTemplate.from_messages([
    ('system','for data science roles find the missing skills from the resume.'),
    ('human','resume: {resume}')
])

missing_skill_chain=missing_skill_prompt | llm | parser
```

```python
final_chain= RunnableParallel(
    skills=skill_chain,
    summarize=summarize_chain,
    missing_skills=missing_skill_chain
)
```
