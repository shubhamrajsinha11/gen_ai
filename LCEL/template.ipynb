```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
```

# LCEL

### step1-> prompt tempelate 

```python
prompt_template=ChatPromptTemplate.from_messages([
    ('human', 'you are a helpful assistant, Detect the language'),
    ('ai', 'Expected Output: {language}')
])
```

### step2-> LLM

```python
lim= ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    api_key= '',
    temperature=0.2
)
```

### step3-> parser

```python
parser= StrOutputParser()
```

```python
language_chain= prompt_template | lim | parser

response= language_chain.invoke('你怎么能对长辈说这么难听的脏话?')
```
