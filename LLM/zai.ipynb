```python
import os
import requests

API_URL = ""
headers = {
    "Authorization": f"Bearer {os.environ['HF_TOKEN']}",
}

def query(payload):
    response = requests.post(API_URL, headers=headers, json=payload)
    return response.json()

response = query({
    "messages": [
        {
            "role": "user",
            "content": "?"
        }
    ],
    "model": "zai-org/GLM-5.2:novita"
})

print(response)
```
