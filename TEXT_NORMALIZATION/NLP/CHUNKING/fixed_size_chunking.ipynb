```python
with open('chunk.txt') as file:
    data=file.read()
    #print(data)
chunks = []
fixed_size=20
for index in range(0,len(data),fixed_size):
    chunk=data[index:index+fixed_size]
    #print(chunk,len(chunk))
    chunks.append(chunk)
print(chunks)
```

```python
with open('chunk.txt') as file:
    data=file.read()
    #print(data)
chunks = []
fixed_size=20
overlapping=4
for index in range(0,len(data),fixed_size-overlapping):
    chunk=data[index:index+fixed_size]
    #print(chunk,len(chunk))
    chunks.append(chunk)
print(chunks)
```

```python
with open('chunk.txt') as file:
    data=file.read()
    #print(data)
chunks = []
fixed_size=20
overlapping=4
paragraph=200
for index in range(0,len(data),paragraph):
    chunk=data[index:index+paragraph]
    #print(chunk,len(chunk))
    chunks.append(chunk)
print(chunks)
```
