```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
word_tokenize('ling nath tiwari')
```

```python
word_tokens=word_tokenize('ling nath tiwari')
word_tokens
```

```python
res= '''To create a paragraph that utilizes the standard set of English punctuation marks, one must carefully integrate periods, commas, question marks, exclamation points, colons, semicolons, dashes, parentheses, brackets, apostrophes, quotation marks, and ellipses. The following paragraph demonstrates this usage: {0}'''
```

```python
res.split()
```

```python
sent_tokens=sent_tokenize(res)
sent_tokens
```

```python
import re
# re.findall(r'\s+', res)
re.sub(r'\s+','',res)
```

```python
re.sub(r'\s{2,}','',res) # this pattern will remove all the whitespace characters from the string, leaving only punctuation and alphanumeric characters.
```

```python
re.sub(r'[a-zA-Z0-9]','',res) # this pattern will remove all the alphanumeric characters from the string, leaving only punctuation and whitespace.
```

```python
re.sub(r'[^a-zA-Z0-9]',' ',res) # this pattern will replace all non-alphanumeric characters with a space.
```
